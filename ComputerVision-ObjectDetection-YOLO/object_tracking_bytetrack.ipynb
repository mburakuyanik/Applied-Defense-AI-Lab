```python
from ultralytics import YOLO

# 1. Modeli yüklüyoruz
model = YOLO("yolov8n.pt")

# 2.Modeli 'Track' modunda çalıştırıyoruz (ByteTrack Algorithm)
# persist=True: Kareler arası ID hafızasını korur
results = model.track(
    source = "https://ultralytics.com/images/bus.jpg",
    persist=True,
    tracker="bytetrack.yaml",
    show=True,
    save=True
)

# 3. Takip Çıktılarını ve Atanan ID'leri İnceleme
for result in results:
    boxes = result.boxes
    if boxes.id is not None:
        print("boxes.id:", boxes.id),
        print("boxes type:", type(boxes))
        track_ids = boxes.id.int().cpu().tolist()
        classes = boxes.cls.int().cpu().tolist()

        for track_id, cls in zip(track_ids, classes):
            name = model.names[cls]
            print(f"Takip Edilen Nesne: {name} ---> Atanan Sabit ID: #{track_id}")
```
