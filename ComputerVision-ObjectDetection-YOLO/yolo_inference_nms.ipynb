```python
from ultralytics import YOLO

# 1. Hazır eğitilmiş hafif YOLO modelini yüklüyoruz (yolov8n = nano/hızlı)
model = YOLO('yolov8n.pt')

# 2. Model ile Çıkarım (Inference) Yapma
# confidence değeri = 0.5 Güveni %50 altındaki nesneleri ele
# IoU = 0.7 : NMS aşamasında çakışan kutuları temizleme eşiği
results = model.predict(
    source="https://ultralytics.com/images/bus.jpg",
    conf=0.5,
    iou=0.7,
    save= True
)

# 3. Tespit Edilen Kutuları İnceleme
print("results : ", dir(results[0]))
print("boxes :", dir(results[0].boxes))

for result in results:
    boxes = result.boxes
    for box in boxes:
        cls_id = int(box.cls[0])
        class_name = result.names[cls_id]
        confidence = float(box.conf[0])
        coords = box.xyxy[0].tolist()
        print(f"Class: {class_name}, Confidence: {confidence}, Coordinates: {coords}")

```
