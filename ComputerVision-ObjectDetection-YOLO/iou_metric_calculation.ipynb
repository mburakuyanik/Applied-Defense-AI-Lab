```python
def calculate_iou(boxA,boxB):
   #box format'ı : [Xmin, Ymin, Xmax, Ymax]
   #1. kesişim alanın kordinatlarını bul
   #2. birleşim alanın kordinatları bul
   #3. kesişim/birleşim alanı hesapla IoU bul
   #4. eğer 0.5 üzerindeyse true positive işaretle
   xA = max(boxA[0], boxB[0])
   yA = max(boxA[1], boxB[1])
   yB = min(boxA[2], boxB[2])
   xB = min(boxA[3], boxB[3])

   interArea = max(0, yB-yA) * max(0, xB-xA)
   boxAArea = (boxA[2] - boxA[0]) * (boxA[3] - boxA[1])
   boxBArea = (boxB[2] - boxB[0]) * (boxB[3] - boxB[1])

   unionArea = float(boxAArea + boxBArea - interArea)
   iou = interArea/unionArea
   return iou

ground_truth = [10,10,90,90]
prediction = [15,10,100,80]

skor = calculate_iou(ground_truth,prediction)
print(f"Hesaplanan IoU skoru: {skor:.4f}")

if skor >= 0.5:
  print("Sonuç: BAŞARILI TESPİT (True Positive)")
else:
  print("Sonuç: YETERSİZ ÇAKIŞMA (False Positive)")
```
