```python
import json

#(TOOLS)

def get_target_distance(target_id: str) -> float:
    """Verilen hedef kimliğine göre radar menzilini (km) döner"""
    radar_db = {"HEDEF-A" : 6.5, "HEDEF-B" : 9}
    return radar_db.get(target_id, 0.0)

def is_within_cirit_range(distance: float) -> bool:
    """CIRIT füzesinin maksimum menzili 8 km'dir."""
    return distance <= 8.0

#Kullanıcı sorusu sonrası json oluşturma ve tool kullanımı

llm_action = {
    "tool" : "get_target_distance",
    "params" : { "target_id": "HEDEF-B" }
}

if llm_action["tool"] == "get_target_distance":
    target_id = llm_action["params"]["target_id"]
    distance = get_target_distance(target_id)
    print(f"[Ajan Log]: Radar mesafesi ölçüldü -> {distance} km")

    in_range = is_within_cirit_range(distance)

    if in_range:
        print(f"[Ajan Yanıtı]: {llm_action['params']['target_id']} {distance} km, CIRIT menzili (8 km) içerisindedir. Atış gerçekleştirilebilir.")

    else:
        print(f"[Ajan Yanıtı]: {llm_action['params']['target_id']} {distance} km, CIRIT menzili (8 km) içerisinde değildir. Atış gerçekleştirilemez.")
```
