```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

document = """
ROKETSAN tarafından geliştirilen CIRIT, 2.75 inç Lazer Güdümlü Füzedir.
Hafif zırhlı/zırhsız araçlara karşı yüksek etkinliğe sahiptir.
Azamî menzili 8 km, asgarî menzili ise 1.5 km olarak belirtilmiştir.
Helikopterler, İHA'lar ve kara araçlarından ateşlenebilmektedir.
"""

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=40,
    chunk_overlap=20,
)

chunks = text_splitter.split_text(document)

for i, chunk in enumerate(chunks):
  print(f"--- Parça {i+1} (Uzunluk: {len(chunk)})---")
  print(chunk)
```
