```python
from chromadb.utils import embedding_functions
import chromadb

sentence_transformer_ef = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

chroma_client = chromadb.Client()
collection = chroma_client.create_collection(
    name="roketsan_docs4",
    embedding_function=sentence_transformer_ef
)

collection.add(
    documents=[
        "CIRIT, helikopterlerden atılan lazer güdümlü bir füzedir.",
        "SOM, uzun menzilli stand-off seyir füzesidir.",
        "ATMACA, gemisavar füze sistemidir.",
        "Ankara'da hava sıcaklığı bugün 32 derecedir.", # dummy data
    ],
    ids=["doc1", "doc2", "doc3", "doc4"]
)

query = "Helikopter silahları nelerdir?"

results = collection.query (
    query_texts=[query],
    n_results=1 #En yakın 1 sonucu getir
)

print(f"Soru: {query}\n")
print("Vector DB'den Gelen En Alakalı Belgeler:")
for doc in results['documents'][0]:
    print(f"- {doc}")
```
