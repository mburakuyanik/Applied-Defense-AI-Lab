```python
import numpy as np

# Durum Vektörü: [Konum (x),
#                 Hız   (v)]
x = np.array([[0.0],
              [10.0]])

# 100 ms örnekleme
dt = 0.1

# Hata Kovaryans Matrisi (Belirsizlik)
P = np.eye(2) * 5.0

# Durum Geçiş Matrisi (Fizik: x_yeni = x + v*dt, v_yeni = v)
F = np.array([[1,dt],
              [0,1]])

# Süreç Gürültüsü (Model sapması)
Q = np.array([[0.01,0],
              [0,0.1]])

# Ölçüm Matrisi: Kameradan sadece HIZ (v) okuyoruz
H_camera = np.array([[0.0,1.0]])
R_camera = np.array([[0.5]]) # kamera hızı gürültüsü

# 1 Adımlık Predict (IMU/Fizik) ve Update (Kamera) Döngüsü
# 1. TAHMİN (PREDICT)
x_pred = np.dot(F, x)
P_pred = np.dot(np.dot(F, P), F.T) + Q

# 2. GÜNCELLEME (UPDATE - Kameradan gelen hız: 10.8 m/s)
z_camera = np.array([[10.8]])
y = z_camera - np.dot(H_camera, x_pred) # İnovasyon (Ölçüm Farkı)
S = np.dot(np.dot(H_camera, P_pred), H_camera.T) + R_camera
K = np.dot(np.dot(P_pred, H_camera.T), np.linalg.inv(S)) # Kalman Kazancı

x_updated = x_pred + np.dot(K, y)
P_updated = P_pred - np.dot(np.dot(K, H_camera), P_pred)

print(f"Düzeltilmiş Konum (x): {x_updated[0, 0]:.2f} metre")
print(f"Düzeltilmiş Hız (v): {x_updated[1, 0]:.2f} m/s")
```
