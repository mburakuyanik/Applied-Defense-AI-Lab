```python
import numpy as np
import matplotlib.pyplot as plt

# Rastgelelik sabitlensin
np.random.seed(42)

time_steps = 50
# 1. 0'dan 100'e 50 adım oluştur (Hangi fonksiyonu kullanıyorduk?)
gercek_konum = np.linspace(0, 100, time_steps)

# 2. Gürültü ekle (Ortalama 0, Standart Sapma 5 olsun)
gurultu = np.random.normal(0, 5, size=time_steps)

# 3. Gerçek konuma gürültüyü ekle
olcum_verisi = gercek_konum + gurultu

print("Veriler oluşturuldu, listenin ilk 5 elemanı:", olcum_verisi[:5])
```

```python
x_tahmin = 0.0
P = 1.0
Q = 0.05
R = 25.0
kalman_sonuclari = []

for z in olcum_verisi:
    x_tahmin = x_tahmin + (100.0 / 50)
    P = P + Q

    K = P / (P + R)

    x_tahmin = x_tahmin + K * (z - x_tahmin)
    P = (1 - K) * P

    kalman_sonuclari.append(x_tahmin)

print("Filtreleme bitti, toplam sonuç sayısı:", len(kalman_sonuclari))
```

```python
plt.figure(figsize=(12, 7))

plt.plot(gercek_konum, label='Gerçek Konum (Hedef)', color='green', linestyle='--', linewidth=2)

plt.scatter(range(len(olcum_verisi)), olcum_verisi, label='Gürültülü Ölçümler', color='red', alpha=0.5, s=20)

plt.plot(kalman_sonuclari, label='Kalman Filtresi Tahmini', color='blue', linewidth=2.5)

plt.title("Kalman Filtresi: Gerçek vs Ölçüm vs Tahmin")
plt.xlabel("Zaman Adımı")
plt.ylabel("Konum (Metre)")
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.show()
```
