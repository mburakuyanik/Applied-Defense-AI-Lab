```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.export(format="onnx", dynamic= True)
print("model başarıyla 'yolov8n.onnx' olarak optimize edildi")

import numpy as np

# Normal çalışan bir motorun sıcaklık telemetrisi (Ortalama 60°C)
np.random.seed(42)
normal_sicaklik = np.random.normal(loc=60.0, scale=2.0, size=20)

# Araya sızan anomali / saldırı verileri ekliyoruz
telemetri_akisi = list(normal_sicaklik)
telemetri_akisi[10] = 85.0  # Anomali 1: Aşırı Isınma
telemetri_akisi[15] = 20.0  # Anomali 2: Sensör Kopması / Sahte Veri

# İstatistiksel Analiz (Z-Score Hesabı)
ortalama = np.mean(telemetri_akisi)
std_sapma = np.std(telemetri_akisi)
esik_degeri = 2.5 # 2.5 standart sapma dışı anomali

print(f"\nTelemetri Ortalaması: {ortalama:.2f}°C, Standart Sapma: {std_sapma:.2f}")
for i, deger in enumerate(telemetri_akisi):
    z_score = abs((deger - ortalama) / std_sapma)
    if z_score > esik_degeri:
        print(f"Alarm! Zaman #{i}: Ölçülen Değer = {deger:.1f}°C (Z-Score: {z_score:.2f}) -> ANOMALİ TESPİT EDİLDİ!")
```
